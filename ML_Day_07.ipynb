```python
# Read data from tsv file

pd.read_csv('dataset_name.tsv', sep='\t', names=['columns name as an array'])  # sep parameter defines it is a tab separated data not comma and names parameters used if there's no column name in my dataset
```

```python
# Index col parameter
# Sometimes there are some unnecessary column which is not relevant so we can tranform that column into our index column
pd.read_csv('dataset_name.csv', index_col='irrelavant_column_name')
```

```python
# Header parameter
# Sometimes some dataset are assigned a number like zero before it's column name row so pandas read it as a row. If we want to read it as our actual colum form then,
pd.read_csv('dataset_name.csv', header=1)
```

```python
# use_cols parameter
# If we know which columns exactly we need from a dataset then we can just read our dataset with those columns just.
pd.read_csv('dataset_name.csv', usecols=['column_name_1', 'column_name_2', 'column_name_3'])
```

```python
# Squeez parameters
# If we only need one column and want to make it a pandas series then we can do it
pd.read_csv('dataset_name.csv', usecols=['column_name'], squeeze=True)
```

```python
# Skiprows/nrows parameters
# If we want to skip any rows
pd.read_csv('dataset_name.csv',skiprows=[0,5])
# We restrict how many rows we need
pd.read_csv('dataset_name.csv',nrows=100)
```

```python
# Encoding parameters
# There will be some dataset where some encoding will be provided.
pd.read_csv('dataset_name.csv', encoding='latin-1')  # There can be other encoding parameters instead of latin if that the case we have to change it to latin then using a code editor
```

```python
# Skip bad lines
# Some dataset contain some bad rows like there will be more record than the column
pd.read_csv('dataset_name.csv', error_bad_lines=False)
```

```python
# dtype parameter
# If we want to convert any column datatype to a different datatype
pd.read_csv('dataset_name.csv', dtype={'column_name':'float64'})
```

```python
# Handling dates
# Dates are store many times as string which cause trouble to read it as date
pd.read_csv('dataset_name.csv', parse_dates=['column_name'])
```

```python
# na_values parameter
# If we want to specify which values we will consider as empty or NaN then we can define it when reading the dataset
pd.read_csv('dataset_name.csv', na_values=['values like hiphen or space etc.'])
```

```python
# Loading a huge dataset into chunks
pd.read_csv('dataset_name.csv', chunksize=10000) # every chunk is a dataframe here
```
