```python
'''
Machine Learning Life Cycle(MDLC)
- It is a set of guidelines that we follow when we develop a machine learning related product. It helps us to develop the product in a easy and efficient way.

There are 9 steps we have to follow-
  1. Frame the problem
  - Define the what is the problem we are going to solve. We have to decide what we have to solve, will the product be offline or online, how much will the cost, who are the customers, is it supervised or unsupervise, where will be the data come form, which algorithm we are going to solve ect.
  2. Gathering the data
  - After defining the problem we have to gather the related and very specific data to solve our problem. We can fetch data by API call or apply web scrapping. So data fetch and store in the right form is crucial.
  3. Data preprocessing
  - When we fetch data the data could be noise, unstructured, unclean. So we need to clean and preprocess data to start our work so that we can get maximum accuracy and the work is exact (Remove missing values, outliers, duplicates etc). We have to transform our data in a format so that our algorithm can work its best on it.
  4. Exploratory Data Analysis(EDA)
  - In this step we actually analyze the data. We try to study the relationship between input and output in the data. We have to know the data so we have to study data here, make graphs, analyze statistics etc.(Outlier detection, unvariate or bivariate, correlation etc.)
  5. Feature engineering and selection
  - Features are inputs. Outputs are dependent on features. Sometimes we need some features which we don't have directly in our dataset. So we need to create new features based on existing features for our work which is a part of features engineering. We need to select actual features and remove unnecessar features here.
  6. Model training, evaluation and selection
  - We actually train the many models on our data and evaluate which is best and which we are going to use for our product. After that we select the model that we need. After selecting model we tune the features so that we get best outputs. This is also called hyper-parameter tuning.
  7. Model deployment
  - We convert our model in a software so users can use the product easily. We pick the model and convert it in a binary file then we convert that binary file into an API. When we give input to our product in the backend it goes to our program and call that API. That API will then give us the prediction as output.
  8. Testing
  - We actually do beta testing. Where we give our model to some loyal customers and take feedback from them.
  9. Optimize
  - After passing the testing phase we deploy our product live for all users. Before launching we do some optimization like create a backup of our model/product, data backup, loadbalancinng, retrain frequency fixing for our model etc.


'''
```

```python
'''
Data Engineer vs Data Scientist vs Data Analyst vs ML Engineer

1. Data Engineer
- Data engineering means gather the data and make them available for others so that next step of MLDLC can work on it. It's not easy to collect the data. There are many steps like web scraping, API calling etc for collecting data. If there's any running website then we don't want to harm the actual DB. So, Data engineer make a Data Warehouse from the DB which we say converting OLTP to OLAP. Build data pipelines(APIs for easy access the data). Handle database and data warehouse. It is actually the second step of MLDLC.
2. Data Analyst
- This role actually analyze, clean and make report about the data. So data analyst mainly give a brief to the next step of MLDLC about the data. It is actullay the third step of MLDLC. Cleaning and organizing raw data, analyzing data to derive insights, creating data visualization, producing and maintaing reports, collaborating with teams, optimizing data collection procedures.
3. Data Scientist
- A data scientist is someone who is better at statistics than any software engineer and better at software engineering than any statistician. EDA phase of MLDLC but a data scientist should have all the skills of data analyst and data engineer. It is like a full package of all the role. The main difference of Data Analyst and Data Scientist is Data analyst analyze the past relation of data where data scientist understand the data and create somethig for future with the data. Best model creation should be the goal of a data scientist.
4. ML Engineer
- The main role ML engineer do is deploy the created and selected model. Integrating the model with the actual software. Deploying ML models to production ready environment, scaling and optimizing the model for production, monitorng and maintanance fo deployed model.



'''
```
