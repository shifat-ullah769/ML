```python
# Encoding Categorical Data
```

```python
'''
Encoding categorical data is the process of converting text-based or categorical variables into a numerical format so that machine learning algorithms can understand and process them.

The Two Types of Categorical Data-
  1. Nominal Data: Categories with no inherent order or ranking.
  2. Ordinal Data: Categories with a natural order or ranking.

Types of encoding-
  1. Ordinal encoding(used on ordinal data)
    - used when input features contains ordinal categorical data with a natural order or ranking.
  2. Label encoding(similar to label encoding)
    - used when output/target features are categorical.
  3. One-hot encoding(Used on nominal data)
    - In this technique we take the types of categories separately and made a vector for each categories. This is how we do one hot encoding.
    - After one hot encoding we have to do Dummy Variable Trap by removing one column of categories(mostly the first one) to avoid multicolliniarity and represent all the categories in a single column.
    -
'''
```

Ordinal Encoding


```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/customer.csv')
```

```python
df.sample(5)
```

```python
df = df.iloc[:,2:]
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.2)
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
X_train
```

```python
oe = OrdinalEncoder(categories=[['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']])
oe.fit(X_train)
```

```python
X_train = oe.transform(X_train)
X_test = oe.transform(X_test)
```

```python
X_train
```

```python
oe.categories_
```
