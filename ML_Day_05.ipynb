```python
'''
Tensors-
  Tensors are mainly a data structure. In machine learning, tensors are multi-dimensional arrays used to store, represent, and process data. Tensors are nothing but a cointainer for numbers/data.
  We can think of a tensor as the ultimate upgrade to standard numbers, vectors, and tables, scaled up to any number of dimensions. Frameworks like PyTorch and TensorFlow use tensors as their core data building blocks.

Dimensions of Tensors (Rank)
  The number of dimensions in a tensor is called its rank or axis:
  1) 0D Tensor (Scalar): A single number.
  Example: Temperature (31.2) or a learning rate (0.001).
  2) 1D Tensor (Vector): A list or array of numbers.
  Example: A word embedding for NLP ([0.12, -0.53, 0.91]).
  3) 2D Tensor (Matrix): A grid or table of numbers with rows and columns.
  Example: A grayscale image of pixels arranged in width and height.
  4) 3D Tensor and Higher (True Tensors): Cubes or multi-axis volumes of numbers.
  Example: A color image has height, width, and three color channels (RGB), making it a 3D tensor. A video adds a fourth dimension for time (frames), making it a 4D tensor.




'''
```
