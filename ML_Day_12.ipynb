```python
# Bi-varialte and Multi-variate Analysis
```

```python
'''
Bivariate Analysis-
  Bivariate analysis is a statistical method used to examine the relationship, strength, and pattern between two different variables at the same time.
  Unlike univariate analysis, which describes only a single variable, bivariate analysis looks for associations, causes, or correlations between an independent variable (X) and a dependent variable (Y).

Multi-variate Analysis-
  Multi-variate analysis is a statistical method used to examine the relationship, strength, and pattern between more than two different variables at the same time.
  Unlike simple reports that look at one factor at a time, this approach helps you understand complex systems and hidden patterns.

'''
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset reading
tips = sns.load_dataset('tips')
titanic = pd.read_csv('titanic.csv')
flights = sns.load_dataset('flights')
iris = sns.load_dataset('iris')
```

Scatterplot(Numerical - Numerical)
- If two variables(column) has numerical datatype


```python
# Bivariate Analysis
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.show()
```

```python
# Multi-variate Analysis
sns.scatterplot(x='total_bill', y='tip', hue='sex', style = 'smoker', size = 'size', data=tips)
plt.show()
```

Barpot(Numerical - Categorical)
- If one variable(column) has numerical and another has
categorical datatype

```python
# Bivariate analysis
sns.barplot(x='time', y='total_bill', data=tips)
```

```python
# Multi-variate analysis
sns.barplot(x='time', y='total_bill', hue='sex', data=tips)
```

Boxplot (Numerical - Categorical)
- If one column has numerical datatype and another has categorical datatype

```python
# Bivariate analysis
sns.boxplot(x='tip', y='sex', data=tips)
```

```python
# Multi-variate analysis
sns.boxplot(x='tip', y='sex', hue='smoker', data=tips)
```

Distplot (Numerical - Categorical)
- If one column has numerical and another has categorical datatype

```python
# Bivariate analysis
sns.distplot(tips[tips['sex'] == 0]['tip'])
sns.distplot(tips[tips['sex'] == 1]['tip'])
```

Heatmap(Categorical - Categorical)
- If both column has categorical datatype

```python
# Bivariate analysis
sns.heatmap(pd.crosstab(tips['sex'], tips['smoker']))
```

```python
# print(pd.crosstab(tips['sex'], tips['smoker']))

# Convert 'Yes' to 1 and 'No' to 0, cast to float, and then compute the mean percentage
print(tips.assign(is_smoker=tips['smoker'].map({'Yes': 1, 'No': 0}).astype(float)).groupby('sex', observed=False)['is_smoker'].mean() * 100)
```

```python
(tips.assign(is_smoker=tips['smoker'].map({'Yes': 1, 'No': 0}).astype(float)).groupby('sex', observed=False)['is_smoker'].mean() * 100).plot(kind='bar')
```

Clustermap(Categorical - Categorical)
- If both column has categorical datatype

```python
# Bivariate analysis
pd.crosstab(tips['sex'], tips['smoker'])
```

```python
sns.clustermap(pd.crosstab(tips['sex'], tips['smoker']))
```

Pairplot
- Collection of scatter plot where multiple numerical columns can be considerd

```python
# Multi-variate/bivariate analysis
sns.pairplot(tips)
sns.pairplot(iris, hue='species')
```

Lineplot(Numerical - Numerical)
- Special case of scatter plot. Combining all the points of scatter plot. If x axis data is time related then lineplot has been used

```python

f = flights.groupby('year').sum(numeric_only=True).reset_index()
print(f)
```

```python
sns.lineplot(x=f['year'], y=f['passengers'])
plt.show()
```

```python
sns.heatmap(flights.pivot_table(values='passengers', index='month', columns='year'))
```
