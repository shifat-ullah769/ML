```python
# EDA using Univariate Analysis
'''
Univariate analysis means analyze the data of only one variable at a time.
Thera are two types of data.
  1)Numerical data
  2)Categorical data
'''
```

```python
import pandas as pd
df = pd.read_csv('train.csv')
df.head()
```

1. Categorical Data

```python
# Countplot
import seaborn as sns
sns.countplot(x='Survived',data=df)
df['Survived'].value_counts()
# df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x='Pclass', data=df)
```

```python
df['Sex'].value_counts()
df['Sex'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x='Embarked', data=df)
```

```python
# Piechart
# If we want to get the same information of bar chart as percentage form.

df['Survived'].value_counts().plot(kind='pie', autopct='%1.2f%%')

```

```python
df['Pclass'].value_counts().plot(kind='pie', autopct='%.2f%%')
```

2.Numerical Data

```python
# Histogram
import matplotlib.pyplot as plt
plt.hist(df['Age'], bins=4)
```

```python
# Distplot
sns.distplot(df['Age'])
# It is just like a PDF(Probability Density Function), which tells us what is the probablity of a random number to be.
```

```python
# Boxplot
sns.boxplot(x=df['Age'])
```

```python
df['Fare'].skew()
```
